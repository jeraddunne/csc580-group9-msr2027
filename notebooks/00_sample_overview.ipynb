# 00 Sample overview

First look at the GitSkills and SpecMine samples through the `msr_pipeline` package.
Run `python scripts/download_samples.py` first. Outputs are cleared before commit.

```python
import pandas as pd

from msr_pipeline import explore
from msr_pipeline.config import get_paths, gitskills_db_path, specmine_dir
from msr_pipeline.load import load_gitskills, load_specmine

pd.set_option("display.max_columns", 40)
paths = get_paths()
paths
```

## GitSkills

```python
gs = load_gitskills(gitskills_db_path(paths))
for name, frame in gs:
    print(f"{name:18s} {len(frame):>8,} rows  {len(frame.columns):>3} cols")
gs.artifacts.head()
```

```python
explore.gitskills_location_summary(gs.artifacts)
```

```python
explore.gitskills_copy_distribution(gs.artifacts).head(15)
```

```python
explore.gitskills_language_summary(gs.artifacts, gs.repos)
```

## SpecMine

```python
sm = load_specmine(specmine_dir(paths))
for name, frame in sm.items():
    print(f"{name:24s} {len(frame):>8,} rows  {len(frame.columns):>3} cols")
sm["spec_files"][["repo_name", "file_path", "spec_tool", "spec_role", "file_total_commits"]].head()
```

```python
explore.specmine_tool_summary(sm["spec_files"])
```

```python
explore.specmine_feature_summary(sm["spec_content_features"])
```

```python
explore.specmine_pr_code_cochange(sm["pull_requests"], sm["pr_files"])
```
